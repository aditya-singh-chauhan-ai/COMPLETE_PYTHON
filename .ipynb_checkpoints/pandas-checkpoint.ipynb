```python
df_cus = pd.DataFrame({
    'Name': ['adii', 'abhay','aman','akash'],
    'age': [10,20,20,10],
    'sal': [100000,200000,15646,16464]
})
df_orders = pd.DataFrame({
    'Name':['Ram','Shyam','Ganshayam'],
    'age':[10,20,30],
    'city':['delhi','mumbai','nagpur']
})
df_merged = pd.merge(df_cus,df_orders, on='age',how='inner')
print('inner join')
print(df_merged)
```
