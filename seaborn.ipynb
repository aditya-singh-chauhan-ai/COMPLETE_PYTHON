```python
import seaborn as sns
import pandas as pd
import matplotlib.pylot as plt
```

```python
x=[1,2,3,4,5,6,7]
y=[20,40,30,40,30,50,30]
df = pd.DataFrame({'Days':x, 'No of People':y})
df.head(7)
sns.lineplot(x='Days',y='No of People', data= df)
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
sns.lineplot(x='day',y='tip',data=df)
```

```python
!pip install matplotlib
import matplotlib.pyplot as plt
sns.lineplot(x='day',y='total_bill',data=df)
plt.show()
```

```python
sns.lineplot(x='day',y='total_bill',data=df, hue='sex',style='time')
plt.show()
```

```python
penguins = sns.load_dataset('penguins')
penguins.head()
```

```python
penguins['species'].value_counts()
```

```python
penguins['island'].value_counts()
```

```python
sns.scatterplot(data = penguins,x='flipper_length_mm',y='body_mass_g')
plt.show()
```

```python
sns.scatterplot(data = penguins,x='flipper_length_mm',y='body_mass_g',hue='island')
plt.show()
```

```python
sns.set_style('ticks')
sns.scatterplot(data = penguins,x='flipper_length_mm',y='body_mass_g',hue='island')
```

```python
sns.set_style('whitegrid')
sns.scatterplot(data = penguins,x='flipper_length_mm',y='body_mass_g',hue='island')
plt.show()
```

```python
sns.set_style('darkgrid')
sns.scatterplot(data = penguins,x='flipper_length_mm',y='body_mass_g',hue='island')
plt.show()
```
